```python
import ipywidgets as widgets
from IPython.display import display, clear_output

# datos
PRECIOS_PRODUCTOS = {
    "Arroz": 50.0,
    "Manzana": 4.0,
    "Uvas": 8.0,
    "Fresa": 2.0,
    "Kiwi": 3.0,
}


lista_productos = []

titulo = widgets.HTML("<h2>REGISTRO DE COMPRAS (Precios Fijos)</h2>")


dropdown_producto = widgets.Dropdown(
    options=list(PRECIOS_PRODUCTOS.keys()),
    description="Producto:",
    disabled=False,
)


lbl_precio = widgets.HTML(
    value=f"<b>Precio Unitario:</b> {PRECIOS_PRODUCTOS[dropdown_producto.value]:.2f} Bs"
)


txt_cantidad = widgets.IntText(description="Cantidad:", value=1)



def cambiar_producto(change):
  lbl_precio.value = (
      f"<b>Precio Unitario:</b> {PRECIOS_PRODUCTOS[change['new']]:.2f} Bs"
  )


dropdown_producto.observe(cambiar_producto, names="value")


btn_registrar = widgets.Button(
    description="Registrar", button_style="success", icon="check"
)

btn_limpiar = widgets.Button(
    description="Vaciar Carrito", button_style="danger", icon="trash"
)


salida = widgets.Output()


def registrar_producto(b):
  with salida:
    clear_output(wait=True)
    producto = dropdown_producto.value
    precio = PRECIOS_PRODUCTOS[producto]
    cantidad = txt_cantidad.value

    if cantidad <= 0:
      print(" Error: La cantidad debe ser mayor a 0.")
      mostrar_resumen()
      return

    total_item = precio * cantidad


    lista_productos.append({
        "nombre": producto,
        "precio": precio,
        "cantidad": cantidad,
        "total": total_item,
    })


    txt_cantidad.value = 1

    print(f"¡{producto} añadido al carrito!")
    mostrar_resumen()



def limpiar_carrito(b):
  with salida:
    clear_output(wait=True)
    lista_productos.clear()
    print(" El carrito ha sido vaciado.")
    mostrar_resumen()


def mostrar_resumen():
  print("\n=== RESUMEN DE LA COMPRA ===")
  if not lista_productos:
    print("El carrito está vacío.")
    return

  gran_total = 0
  for i, item in enumerate(lista_productos, 1):
    print(
        f"{i}. {item['nombre']} | {item['cantidad']} x {item['precio']:.2f} Bs"
        f" = {item['total']:.2f} Bs"
    )
    gran_total += item["total"]

  print("-" * 40)
  print(f"TOTAL A PAGAR: {gran_total:.2f} Bs")



btn_registrar.on_click(registrar_producto)
btn_limpiar.on_click(limpiar_carrito)


botones = widgets.HBox([btn_registrar, btn_limpiar])

display(
    titulo, dropdown_producto, lbl_precio, txt_cantidad, botones, salida
)
```
