```python
import ipywidgets as widgets
from IPython.display import display
import pandas as pd
import os
from datetime import datetime

# ============================
# CONFIGURACIÓN
# ============================

ARCHIVO = "productos.csv"

columnas = [
    "Producto",
    "Precio Unitario (Bs)",
    "Cantidad",
    "Total (Bs)",
]

# Si ya existe el archivo, cargar los productos anteriores
if os.path.exists(ARCHIVO):
    productos = pd.read_csv(ARCHIVO)
else:
    productos = pd.DataFrame(columns=columnas)


# ============================
# INTERFAZ
# ============================

titulo = widgets.HTML(
    """
    <h2 style="color:#2E86C1;">
        REGISTRO DE PRODUCTOS
    </h2>
    """
)

txt_producto = widgets.Text(
    description="Producto:",
    placeholder="Escriba el nombre"
)

txt_precio = widgets.FloatText(
    description="Precio (Bs):",
    value=0.0
)

txt_cantidad = widgets.IntText(
    description="Cantidad:",
    value=1
)

btn_registrar = widgets.Button(
    description="Registrar producto",
    button_style="success",
    icon="check"
)

salida_mensaje = widgets.Output()

salida_tabla = widgets.Output()


# ============================
# MOSTRAR TABLA
# ============================

def mostrar_productos():
    with salida_tabla:

        salida_tabla.clear_output(wait=True)

        if productos.empty:
            print("Todavía no existen productos registrados.")
        else:
            display(productos)


# ============================
# REGISTRAR PRODUCTO
# ============================

def registrar_producto(b):

    global productos

    producto = txt_producto.value.strip()
    precio = txt_precio.value
    cantidad = txt_cantidad.value

    # Validaciones
    if not producto:

        with salida_mensaje:
            salida_mensaje.clear_output()
            print("❌ Error: escriba el nombre del producto.")

        return

    if precio < 0:

        with salida_mensaje:
            salida_mensaje.clear_output()
            print("❌ Error: el precio no puede ser negativo.")

        return

    if cantidad <= 0:

        with salida_mensaje:
            salida_mensaje.clear_output()
            print("❌ Error: la cantidad debe ser mayor a 0.")

        return

    # Calcular total
    total = precio * cantidad

    # Fecha y hora
    fecha = datetime.now().strftime("%d/%m/%Y %H:%M:%S")

    # Nuevo registro
    nuevo_producto = pd.DataFrame([{
        "Producto": producto,
        "Precio Unitario (Bs)": precio,
        "Cantidad": cantidad,
        "Total (Bs)": total,
    }])

    # AGREGAR el producto sin borrar los anteriores
    productos = pd.concat(
        [productos, nuevo_producto],
        ignore_index=True
    )

    # Guardar todos los datos en CSV
    productos.to_csv(
        ARCHIVO,
        index=False
    )

    # Mensaje
    with salida_mensaje:

        salida_mensaje.clear_output()

        print("✅ PRODUCTO REGISTRADO CORRECTAMENTE")
        print("-----------------------------------")
        print(f"Producto: {producto}")
        print(f"Precio unitario: {precio:.2f} Bs")
        print(f"Cantidad: {cantidad}")
        print(f"Total: {total:.2f} Bs")

    # Limpiar campos para registrar otro
    txt_producto.value = ""
    txt_precio.value = 0.0
    txt_cantidad.value = 1

    # Actualizar tabla
    mostrar_productos()


# Asociar botón
btn_registrar.on_click(registrar_producto)


# ============================
# MOSTRAR PROGRAMA
# ============================

display(
    titulo,
    txt_producto,
    txt_precio,
    txt_cantidad,
    btn_registrar,
    salida_mensaje,
    widgets.HTML("<h3>Productos registrados</h3>"),
    salida_tabla
)

mostrar_productos()
```
